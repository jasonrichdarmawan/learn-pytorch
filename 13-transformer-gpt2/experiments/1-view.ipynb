```python
import torch

x = torch.arange(2 * 3 * 4).reshape(2, 3, 4)
x2 = x.view(2 * 3, 4)
y = torch.randint(4, (2, 3))
y2 = y.view(2 * 3)
print(f"x:\n{x}")
print(f"x2:\n\{x2}")
print(f"x2[0]:\n{x2}")
print(f"y:\n{y}")
print(f"y:\n{y2}")
```

```python
logits = torch.arange(2 * 3 * 4, dtype=torch.float32).reshape(2, 3, 4)
logits2 = logits[:, -1, :]
print(f"logits:\n{logits}")
print(f"logits2:\n{logits2}")
print(f"logits2.shape: {logits2.shape}")
```

```python
probs = torch.tensor([[0.05, 0.0, 0.0, 0.95],
                      [0.1, 0.2, 0.7, 0.0],
                      [0.2, 0.3, 0.5, 0.0]])

idx_next = torch.multinomial(probs, num_samples=1)

print(f"probs:\n{probs}")
print(f"idx_next:\n{idx_next}")
```

```python
# Create two sample tensors with the same shape
x = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])
y = torch.tensor([[7, 8, 9],
                  [10, 11, 12]])
print(f"x.shape: {x.shape}")
print(f"y.shape: {y.shape}")

# Stack tensors along a new dimension (default dim=0)
stacked_dim0 = torch.stack([x, y])
print("Stacked along dimension 0:")
print(stacked_dim0)
print(f"stacked_dim0.shape: {stacked_dim0.shape}")
```

```python
# Create two sample 2D tensors with the same number of rows
idx = torch.tensor([[1, 2, 3],
                    [4, 5, 6]])
idx_next = torch.tensor([[7],
                         [10]])

# Concatenate t1 and t2 along dimension 1 (columns)
concatenated_dim1 = torch.cat([idx, idx_next], dim=1)
print("Concatenated tensor along dim=1:")
print(concatenated_dim1)
```

```python
# Create two tensors that are close within a tolerance
tensor1 = torch.tensor([1.0, 2.0, 3.0])
tensor2 = torch.tensor([1.0, 2.00005, 2.99995])

# Check if all elements are close with absolute tolerance of 1e-4
result = torch.allclose(tensor1, tensor2, atol=1e-4)
print("Are tensor1 and tensor2 close?", result)
```
