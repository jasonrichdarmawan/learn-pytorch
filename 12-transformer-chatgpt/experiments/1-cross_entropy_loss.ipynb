```python
import torch

loss = torch.nn.CrossEntropyLoss()

input = torch.tensor([1, float('-inf')])
target = torch.tensor(0)
output = loss(input, target)
print(f"input.shape: {input.shape}")
print(f"target.shape: {target.shape}")
print(f"output: {output}")

input = torch.tensor([[1, float('-inf')],
                      [float('-inf'), 1]])
target = torch.tensor([0,
                       1])
output = loss(input, target)
print(f"input.shape: {input.shape}")
print(f"target.shape: {target.shape}")
print(f"output: {output}")

input = torch.tensor([[[            1, float('-inf'), float('-inf')],
                       [float('-inf'),             1, float('-inf')]],
                      [[float('-inf'),             1, float('-inf')],
                       [            1, float('-inf'), float('-inf')]]])
# input: (N, d1, C)
input = input.transpose(1, 2)
# input: (N, C, d1)
target = torch.tensor([[0, 1],
                       [1, 0]])
output = loss(input, target)
# expected input: (N, C, d1)
# expected target: (N, d1)
print(f"input.shape: {input.shape}")
print(f"input:\n{input}")
print(f"target.shape: {target.shape}")
print(f"target:\n{target}")
print(f"output: {output}")
```

```python
import torch

token_to_id = {'what': 0, 'is': 1, 'statquest': 2, 'awesome': 3, '<EOS>': 4}
id_to_token = dict(map(reversed, token_to_id.items()))

n = 2
q_len = 6
num_tokens = 5

output = torch.tensor([[token_to_id[token] for token in sentence.split()]
                       for sentence in ["what is statquest <EOS> awesome awesome",
                                        "what is statquest <EOS> awesome awesome"]],
                       dtype=torch.float32)
# output: (n, q_len, num_tokens)
print(f"output:\n{output}")

labels = torch.tensor([[token_to_id[token] for token in sentence.split()]
                       for sentence in ["what is statquest <EOS> awesome <EOS>",
                                        "what is statquest <EOS> awesome <EOS>"]],
                       dtype=torch.float32)
# labels: (n, q_len)
print(f"labels:\n{labels}")

criterion = torch.nn.CrossEntropyLoss()
loss = criterion(output, labels)
print(f"loss: {loss}")
```
