```python
import torch

import torch.nn as nn

# Define vocabulary sizes and embedding dimensions
vocab_size = 10
embedding_dim = 5

# Create a tensor of indices for lookup
indices = torch.LongTensor([2, 4, 6])

# Initialize an embedding layer
embedding = nn.Embedding(vocab_size, embedding_dim)

# For reproducibility, use a fixed seed and assign random weights to the embedding layer
torch.manual_seed(0)
embedding.weight.data = torch.randn(vocab_size, embedding_dim)

# Lookup using the nn.Embedding layer
emb_output = embedding(indices)

# Manual embedding: convert indices to one-hot vectors and multiply with the weight matrix
one_hot = torch.eye(vocab_size)[indices]

manual_output_matmul = one_hot @ embedding.weight.data
manual_output_indices = embedding.weight.data[indices]

print("Embedding weight:\n", embedding.weight.data)
print("Embedding lookup output:\n", emb_output)
print("One hot vectors:\n", one_hot)
print("Manual multiplication matmul output:\n", manual_output_matmul)
print("Manual multiplication indices output:\n", manual_output_indices)
```

```python
linear = torch.nn.Linear(embedding_dim, vocab_size, bias=False)
linear.weight = embedding.weight
unembed_output = linear(emb_output)
print("Unembedding output:\n", unembed_output)
```
