```python
import torch

# Create a new generator and set a manual seed for reproducibility
gen = torch.Generator()
gen.manual_seed(42)

# Use the generator to produce a random tensor
rand_tensor = torch.randn(3, 3, generator=gen)
print("Random tensor generated with torch.Generator:")
print(rand_tensor)

# Demonstrate reproducibility by generating the same tensor again
gen.manual_seed(42)
rand_tensor_repeat = torch.randn(3, 3, generator=gen)
print("\nReproduced random tensor:")
print(rand_tensor_repeat)
```
