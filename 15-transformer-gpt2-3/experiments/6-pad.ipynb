```python
import torch

import torch.nn.functional as F

# Create a 4D tensor with shape (batch_size, channels, height, width).
# Here, we create a single-channel 3x3 image.
input_tensor = torch.arange(9, dtype=torch.float32).view(1, 1, 3, 3)
print("Original tensor:")
print(input_tensor)

# Use F.pad to pad the tensor.
# The pad argument follows the format (pad_left, pad_right, pad_top, pad_bottom).
# This example pads the image with 1 column on the left and right sides, 
# and 2 rows on the top and bottom.
padded_tensor = F.pad(input_tensor, (1, 2, 3, 4), mode='constant', value=0)
print("\nPadded tensor:")
print(padded_tensor)
```
