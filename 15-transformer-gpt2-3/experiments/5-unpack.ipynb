```python
# Demonstration of unpacking using '*' and the zip() function

# 1. Using '*' to unpack function arguments
def add(a, b, c):
  return a + b + c

numbers = [1, 2, 3]
result = add(*numbers)  # Unpacks the list into the three arguments
print("Sum using unpacking in a function:", result)

# 2. Using '*' to merge lists
list1 = [1, 2, 3]
list2 = [4, 5, 6]
merged_list = [*list1, *list2]  # Unpacks both lists into one list
print("Merged list:", merged_list)

# 3. Using zip() to combine lists (creates list of tuples)
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]
combined = list(zip(names, ages))
print("Combined using zip:", combined)

# 4. Using '*' with zip() to unzip a list of tuples
unzipped_names, unzipped_ages = zip(*combined)
print("Unzipped names:", unzipped_names)
print("Unzipped ages:", unzipped_ages)
```
