# Experiment 1

How Multi-head Attention is used in Transformer.

Note: I don't aim to be right about Multi-head Attention, I just want to do the big picture.

````python
import torch
import torch.nn as nn

N = 3
vocab_size = 4
query_len = 2
embed_size = 4
heads = 2
head_dim = embed_size // heads
assert head_dim * heads == embed_size, "embed_size must be divisible by heads"
print(f"head_dim: {head_dim}")

x = torch.zeros(N, query_len, dtype=torch.long)
print(f"x shape:\n{x.shape}")
print(f"x:\n{x}")

word_embedding = nn.Embedding(vocab_size, embed_size)
x = word_embedding(x)
print(f"x after word_embedding shape:\n{x.shape}")
print(f"x after word_embedding:\n{x}")

linear = nn.Linear(embed_size,embed_size)
y = linear(x)
print(f"y:\n{y}")
print(f"w:\n{linear.weight}")
print(f"b:\n{linear.bias}")

y_heads = y.view(N, query_len, heads, head_dim)
print(f"y_heads:\n{y_heads}")
````
