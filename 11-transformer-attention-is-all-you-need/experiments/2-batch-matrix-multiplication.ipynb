```python
import torch

# Batch matrix multiplication
a = torch.arange(start=1, end=121, step=1).reshape(2,3,4,5) # 2x3x4x5
b = torch.arange(start=1, end=121, step=1).reshape(2,3,5,4) # 2x3x5x4
torch.matmul(a,b).shape # 2x3x4x4
```

```python
a = torch.arange(start=1, end=13, step=1).reshape(1,2,2,3) # 1x2x2x3 N, i, W, C
b = torch.arange(start=1, end=19, step=1).reshape(1,3,2,3) # 1x3x2x3 N, j, W, C
print("a")
print(a)
print("b")
print(b)
einsum = torch.einsum("niwc,njwc->nwic", a, b) # 1x2x2x3 N, W, b, h_2 (sum over j)
M = torch.empty((1,2,2,3))
for n in range(1):
  for w in range(2):
    for i in range(2):
      for c in range(3):
        for j in range(3):
          M[n,w,i,c] += a[n,i,w,c] * b[n,j,w,c]
print(einsum)
print(M)
```

# Convolution

```python
a = torch.arange(start=1, end=25, step=1).reshape(2,2,2,3) # N, H, W, C
b = torch.arange(start=1, end=25, step=1).reshape(2,2,2,3) # N, H, W, C
print("a")
print(a)
print("b")
print(b)
einsum = torch.einsum("nhwc,nhwc->nhw", a, b) # N, H, W, C
M = torch.empty((2,2,2))
for n in range(2):
  for h in range(2):
    for w in range(2):
      for c in range(3):
        M[n,h,w] += a[n,h,w,c] * b[n,h,w,c]
print("einsum")
print(einsum.shape)
print(einsum)
print("M")
print(M.shape)
print(M)
```

```python
# nhwc,nhwc->nhw # sum over C
a = torch.arange(start=1, end=25, step=1).reshape(2,2,2,3) # N, H, W, C
b = torch.arange(start=1, end=25, step=1).reshape(2,2,2,3) # N, H, W, C
M = torch.empty((2,2,2))
n = 0
h = 0
w = 0
for c in range(3):
  M[n, h, w] += a[n,h,w,c] * b[n,h,w,c]
print(M)
```

# Transformer architecture

```python
a = torch.arange(start=1, end=37, step=1).reshape(2,3,2,3) # N, query_len, heads, head_dim
b = torch.arange(start=1, end=49, step=1).reshape(2,4,2,3) # N, key_len, heads, head_dim
print("a")
print(a)
print("b")
print(b)
einsum = torch.einsum("nqhd,nkhd->nhqk", a, b) # 2x2x3x4 N, heads, query_len, key_len
M = torch.empty((2,2,3,4))
for n in range(2):
  for h in range(2):
    for q in range(3):
      for k in range(4):
        for d in range(3):
          M[n,h,q,k] += a[n,q,h,d] * b[n,k,h,d]
print("einsum")
print(einsum.shape)
print(einsum)
print("M")
print(M.shape)
print(M)
```

# Transformer

```python
# key_len and value_len are the same
# attention shape: (N, heads, query_len, key_len)
# values shape: (N, value_len, heads, head_dim)
# out shape: (N, query_len, heads, head_dim)
attention = torch.arange(start=1, end=49, step=1).reshape(2,3,2,4) # N, heads, query_len, key_len
values = torch.arange(start=1, end=73, step=1).reshape(2,4,3,3) # N, value_len, heads, head_dim
print("attention")
print(attention)
print("values")
print(values)
einsum = torch.einsum("nhql,nlhd->nqhd", attention, values) # N, query_len, heads, head_dim
M = torch.empty((2,2,3,3))
for n in range(2):
  for q in range(2):
    for h in range(3):
      for d in range(3):
        for l in range(4):
          M[n,q,h,d] += attention[n,h,q,l] * values[n,l,h,d]
print("einsum")
print(einsum.shape)
print(einsum)
print("M")
print(M.shape)
print(M)
```

```python
# key_len and value_len are the same
# attention shape: (N, heads, query_len, key_len)
# values shape: (N, value_len, heads, head_dim)
# out shape: (N, query_len, heads, head_dim)
attention = torch.arange(start=1, end=49, step=1).reshape(2,3,2,4) # N, heads, query_len, key_len
values = torch.arange(start=1, end=73, step=1).reshape(2,4,3,3) # N, value_len, heads, head_dim
M = torch.empty((2,2,3,3))
n = 0
q = 0
h = 0
d = 0
for l in range(4):
  M[n,q,h,d] += attention[n,h,q,l] * values[n,l,h,d]
print("M")
print(M.shape)
print(M)
```
