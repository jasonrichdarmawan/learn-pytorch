```python
import torch

a = torch.arange(0, 10, step=1)
print("a")
print(a)
print("a.expand(3, 10)")
print(a.expand(3, 10))
```

```python
a = torch.arange(0, 10, step=1).reshape(2, 5)
print(a.shape)
print(a)
a = a.unsqueeze(1)
print(a.shape)
print(a)
a = a.unsqueeze(2)
print(a.shape)
print(a)
```

```python
a = torch.tril(torch.ones(3, 3))
print(a)
print(a.shape)
a = a.expand(2, 1, 3, 3)
print(a)
print(a.shape)
```
