```python
# https://docs.python.org/3/library/typing.html#generics
# https://docs.python.org/3/library/dataclasses.html#inheritance

from dataclasses import dataclass
from typing import TypeVar, Generic

# Use Case 1
# Standard use
@dataclass
class BaseArgs:
  name: str

@dataclass
class BaseSampler:
  name: str

T_Args = TypeVar('T_Args', bound=BaseArgs)
class BaseTrainer(Generic[T_Args]):
  def __init__(self, args: T_Args, sampler: BaseSampler):
    self.args = args
    self.sampler = sampler

  def train(self):
    print("BaseTrainer", self.args.name)
    print("BaseTrainer", self.sampler.name)

args = BaseArgs(name="Base Args name")
sampler = BaseSampler(name="Base Sampler")
trainer = BaseTrainer(args=args, sampler=sampler)
trainer.train()

# Use Case 2
# Want to initialize BaseTrainer with V2Args
@dataclass
class V2Args(BaseArgs):
  description: str

args = V2Args(name="V2 Args name", description="V2 Args description")
sampler = BaseSampler(name="Base Sampler")
trainer = BaseTrainer[V2Args](args=args, sampler=sampler)
print("BaseTrainer[V2Args]", trainer.args.description)

# Use Case 3
# Want to pass BaseTrainer with V2Args as parameter
def function(trainer: BaseTrainer[V2Args]):
  print(trainer.args.description)
function(trainer)

# Use Case 4
# Want to create V2Trainer with V2Args
class V2Trainer(BaseTrainer[V2Args]):
  def train(self):
    print("V2Trainer", self.args.name, self.args.description)
    print("V2Trainer", self.sampler.name)

trainer = V2Trainer(args=args, sampler=sampler)
trainer.train()

# Use Case 5
# Can't edit V2Trainer
# Want to add V2Sampler
@dataclass
class V2Sampler(BaseSampler):
  description: str

class V3Trainer(V2Trainer):
  sampler: V2Sampler

  def __init__(self, args, sampler: V2Sampler):
    super().__init__(args, sampler)

  def train(self):
    print("V3Trainer", self.args.name, self.args.description)
    print("V3Trainer", self.sampler.name, self.sampler.description)

sampler = V2Sampler(name="V2 Sampler name", description="V2 Sampler description")
trainer = V3Trainer(args=args, sampler=sampler)
trainer.train()
```
