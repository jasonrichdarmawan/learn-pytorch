```python
import torch as t
from eindex import eindex

batch_size = 2
seq_len = 3
d_vocab = 4

logprobs = t.arange(batch_size * seq_len * d_vocab).view(batch_size, seq_len, d_vocab)
labels = t.tensor([[0, 1, 2],
                   [1, 2, 3]])

output_1 = eindex(logprobs, labels, "batch seq [batch seq]")
output_2 = t.empty(batch_size, seq_len, dtype=t.int64)
output_3 = logprobs[t.arange(batch_size).unsqueeze(-1), t.arange(seq_len), labels]

print(f"logprobs=\n{logprobs}")
print(f"labels=\n{labels}")
for i in range(batch_size):
  for j in range(seq_len):
    output_2[i, j] = logprobs[i, j, labels[i, j]]
print(f"output_1=\n{output_1}")
print(f"output_2=\n{output_2}")
print(f"output_3=\n{output_3}")
# assert t.allclose(output_1, output_2), "eindex output does not match expected output"
# assert t.allclose(output_1, output_3), "eindex output does not match expected output"
```
