```python
import torch
import torch.nn as nn

class MyDenseLayer(nn.Module):
  def __init__(self, input_dim: int, output_dim: int):
    super().__init__()

    # Initialize weights and bias
    self.W = nn.Parameter(torch.randn(input_dim, output_dim, requires_grad=True))
    self.b = nn.Parameter(torch.randn(1, output_dim), requires_grad=True)
  
  def forward(self, inputs: torch.Tensor):
    # Forward propagate the inputs
    z = torch.matmul(inputs, self.W) + self.b

    # Feed through a non-linear activation
    output = torch.sigmoid(z)
    return output
```

```python
import torch
import torch.nn as nn

def generate_model():
  model = nn.Sequential(
    # e.g. 3x32x32
    nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3),
    # 3x32x32 -> 3x(32-3+1)x(32-3+1) = 32x30x30
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2, stride=2),
    # 32x30x30 -> 32x15x15

    nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3),
    # 32x15x15 -> 64x13x13
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2, stride=2),
    # 64x13x13 -> 64x6x6

    # fully connected classifier
    nn.Flatten(),
    nn.Linear(64*6*6, 1024), # flattened dim after 2 conv layers,
    nn.ReLU(),
    nn.Linear(1024, 10)
  )

  return model

x = torch.randn(1, 3, 32, 32)
model = generate_model()
output = model(x)
```
