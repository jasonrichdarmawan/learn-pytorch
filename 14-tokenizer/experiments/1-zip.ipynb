```python
# Define two iterables
numbers = [1, 2, 3]
letters = ['a', 'b', 'c', 'd']

# Using zip will stop at the shortest iterable (list)
zipped_list = list(zip(numbers, letters))
print("Zipped result:", zipped_list)

# Unzip the zipped list back into two tuples
unzipped_numbers, unzipped_letters = zip(*zipped_list)
print("Unzipped numbers:", unzipped_numbers)
print("Unzipped letters:", unzipped_letters)
```

```python
ids = [4,3,2,1,
       4,3,2,1]

counts = {}

for pair in zip(ids, ids[1:]):
  counts[pair] = counts.get(pair, 0) + 1

stats = counts

print(f"counts:\n{counts}")
print(sorted(stats.items(), reverse=True))
print(list((v,k) for k,v in stats.items()))
print(sorted(((v,k) for k,v in stats.items()), reverse=True))
```
