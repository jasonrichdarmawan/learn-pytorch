Variants:
1. NFD (canonical decomposition)

   Ç (\u00C7) -> Ç (\u0043\u0327)
  
2. NFC (Canonical decomposition followed by canonical composition)

   Ç (\u0043\u0327) -> Ç (\u00C7)

3. NFKD (compatibility decomposition)

   ℋ (\u210B) -> H (\u0048)

   ℋ̧ ("\u210B\u0327") -> Ḩ ("\u0048\u0327")

4. NFKC (compatibility decomposition followed by canonical composition)

   ℋ (\u210B\u0327) -> Ḩ (\u0048\u0327) -> Ḩ (\u1E28)

Reference:
1. [Normalization NFKC](https://www.youtube.com/watch?v=vWzdEbyQ8qw)

```python
import sentencepiece as spm

spm.SentencePieceTrainer.Train(
  sentence_iterator=iter(["Hello world",
                          "你好。",
                          "Let's go"]),
  model_prefix="output/spm_model",
  model_type="bpe",
  vocab_size=31, # 256 fallback + 3 special tokens + 7 single tokens + 3 bpe tokens + 3 no idea
)

sp = spm.SentencePieceProcessor()
sp.Load(model_file="output/spm_model.model")
vocab = [[sp.IdToPiece(idx), idx] for idx in range(sp.GetPieceSize())]
print(vocab) # check the vocab order, it's changed.
print(f"Hello world: {sp.EncodeAsPieces('Hello world')}")
print(f"Hello world: {sp.Encode('Hello world')}")
print(f"Hello world: {sp.Decode(sp.Encode('Hello world'))}")
print(f"你好。: {sp.EncodeAsPieces('你好。')}")
print(f"你好。: {sp.Encode('你好。')}")
print(f"你好。: {sp.Decode(sp.Encode('你好。'))}")
print(f"Let's go:", sp.EncodeAsPieces("Let's go"))
print(f"Let's go:", sp.Encode("Let's go"))
print(f"Let's go:", sp.Decode(sp.Encode("Let's go")))
```

```python
import unicodedata

print(unicodedata.normalize("NFD", "\u00C7") == "\u0043\u0327") # Ç -> C + cedilla
# canonical decomposition
print(unicodedata.normalize("NFC", "\u0043\u0327") == "\u00C7") # C + cedilla -> Ç
# canonical composition

# usage?
print(unicodedata.normalize("NFC", "\u00C7") == unicodedata.normalize("NFC", "\u0043\u0327"))
```

```python
import unicodedata

print(unicodedata.normalize("NFD", "\u210B") == "\u210B")
# canonical decomposition (D), there is nothing to decompose
print(unicodedata.normalize("NFKD", "\u210B\u0327") == "\u0048\u0327")
# compatibility decomposition (KD)
print(unicodedata.normalize("NFKC", "\u210B\u0327") == "\u1E28")
# compatibility decomposition followed by canonical composition (KC)

# usage?
print(unicodedata.normalize("NFKC", "\u210B\u0327") == unicodedata.normalize("NFKC", "\u0048\u0327"))
```
