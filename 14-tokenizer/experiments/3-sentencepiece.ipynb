# how SentencePiece (with BPE) merge token

same.

```python
import sentencepiece as spm

spm.SentencePieceTrainer.Train(
  sentence_iterator=iter(["deep learning engineer"]),
  model_prefix="output/spm_model",
  model_type="bpe",
  vocab_size=13,
)

sp = spm.SentencePieceProcessor()
sp.Load(model_file="output/spm_model.model")
vocab = [[sp.IdToPiece(idx), idx] for idx in range(sp.GetPieceSize())]
print(vocab) # check the vocab order
print(sp.Encode("deep learning engineer"))
```

# byte_Fallback

```python
print(f"J:", 'J'.encode('utf-8'), list(b'J'), bytes([74]).decode('utf-8'))
print(f"안J", '안J'.encode('utf-8'), list(b'\xec\x95\x88J'), bytes([236, 149, 136, 74]).decode('utf-8'))
# str.encode() -> bytes
# bytes.decode() -> str
```

```python
import sentencepiece as spm

spm.SentencePieceTrainer.Train(
  sentence_iterator=iter(["I"]),
  model_prefix="output/spm_model",
  model_type="bpe",
  vocab_size=261,
  byte_fallback=True
)

sp = spm.SentencePieceProcessor()
sp.Load(model_file="output/spm_model.model")
vocab = [[sp.IdToPiece(idx), idx] for idx in range(sp.GetPieceSize())]
print(vocab) # check the vocab order, it's changed.
print(f"J: {sp.EncodeAsPieces('J')}")
print(f"J: {sp.Encode('J')}")
print(f"안J: {sp.EncodeAsPieces('안J')}")
print(f"안J: {sp.Encode('안J')}")
```
